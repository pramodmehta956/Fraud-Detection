```python
from ctransformers import AutoModelForCausalLM

# Load a quantized LLaMA model
llm = AutoModelForCausalLM.from_pretrained(
    "TheBloke/Llama-2-7B-Chat-GGUF",
    model_file="llama-2-7b-chat.Q4_K_M.gguf",  # Adjust to your downloaded GGUF model
    model_type="llama",
    max_new_tokens=256,
    temperature=0.7
)

response = llm("What is the capital of France?")
print(response)

```

```python
from sentence_transformers import SentenceTransformer

model = SentenceTransformer("elastic/multilingual-e5-small")

sentences = [
    "The weather is lovely today.",
    "It's so sunny outside!",
    "He drove to the stadium."
]
embeddings = model.encode(sentences)

similarities = model.similarity(embeddings, embeddings)
print(similarities.shape)
# [3, 3]
```

```python
from fastapi import FastAPI, Request
from pydantic import BaseModel
from ctransformers import AutoModelForCausalLM

# Load GGUF Model
llm = AutoModelForCausalLM.from_pretrained(
    ".",  # Current directory where model is downloaded
    model_file="mistral-7b-instruct-v0.1.Q4_K_M.gguf",
    model_type="mistral",
    max_new_tokens=512,
    temperature=0.7
)

# Define FastAPI app
app = FastAPI()

# Request schema
class PromptRequest(BaseModel):
    prompt: str

@app.post("/generate")
def generate_text(req: PromptRequest):
    output = llm(req.prompt)
    return {"response": output}
```
